```python
import pandas as pd
import numpy as np
```

```python
path_to_file = 'three_years_in_row_OPEX.xlsx'
df = pd.read_excel(path_to_file,sheet_name="Sheet1",  header = 0, index_col = 0)
df_fact = pd.read_excel('./.xlsx', sheet_name = 'Факт', header = 0)
df_plan = pd.read_excel('./.xlsx', sheet_name = 'План', header = 0)
df_frcst = pd.read_excel('./.xlsx', sheet_name = 'Прогноз', header = 1)
```

```python
df_fact.rename(columns = {'     Драйвер \n(загруженный)': 'Факт'}, inplace = True)
df_plan.rename(columns = {'     Драйвер \n(загруженный)': 'План'}, inplace = True)
df_frcst.rename(columns = {'     Драйвер \n(загруженный)': 'Прогноз'}, inplace = True)
```

```python
df['Concat'] = df['item_id'].astype(str) + '$' + df['timestamp'].astype(str)

for col in df_frcst.columns:
    print(repr(col)) 
```

```python
df_fact['Concat'] = df_fact['Поставщик. ЦФО'].astype(str) + '$' + df_fact['Драйвер'].astype(str) + '$' + df_fact['Услуга'].astype(str) + '$' + df_fact['05. Календарный год'].astype(str) + '-' + '0' + df_fact['04. Календарный месяц'].astype(str) + '-' + '01'
df_plan['Concat'] = df_plan['Поставщик. ЦФО'].astype(str) + '$' + df_plan['Драйвер'].astype(str) + '$' + df_plan['Услуга'].astype(str) + '$' + df_plan['05. Календарный год'].astype(str) + '-' + '0' + df_plan['04. Календарный месяц'].astype(str) + '-' + '01'
df_frcst['Concat'] = df_frcst['Поставщик. ЦФО'].astype(str) + '$' + df_frcst['Драйвер'].astype(str) + '$' + df_frcst['Услуга'].astype(str) + '$' + df_frcst['05. Календарный год'].astype(str) + '-' + '0' + df_frcst['04. Календарный месяц'].astype(str) + '-' + '01'

df_frcst[df_frcst['Услуга']== 152773].tail(3)
```

```python
df = df.merge(df_fact[['Concat','Факт']], on = ['Concat'], how = 'left')
df = df.merge(df_plan[['Concat','План']], on = ['Concat'], how = 'left')
df = df.merge(df_frcst[['Concat','Прогноз']], on = ['Concat'], how = 'left')
df = df.drop(columns = ['Concat','0.1','0.9'])
```

```python
df['Дельта'] = df['mean'] - df['Факт']
df['Прирост'] = ((df['mean'] - df['Факт']) / df['Факт']*100).round(0)
df['Дельта_план'] = df['mean'] - df['План']
df['Прирост_план'] = ((df['mean'] - df['План']) / df['План']*100).round(0)
df['Дельта_прогноз'] = df['mean'] - df['Прогноз']
df['Прирост_прогноз'] = ((df['mean'] - df['Прогноз']) / df['Прогноз']*100).round(0)
```

```python
rating = {
    100 : 'Отклонение более 100%',
    50 : 'Отклонение не менее 50%',
    25 : 'Отклонение не более 25%',
    '':'Отклонение в пределах погрешности',
}
def get_rating(value):
    if pd.isna(value):
        return np.nan
    abs_value = abs(value)
    if abs_value >= 100:
        return rating[100]
    elif abs_value >= 50:
        return rating[50]
    elif abs_value >= 25:
        return rating[25]
    else:
        return rating['']

df['Рейтинг'] = df['Прирост'].apply(get_rating)

precision_rows = df[df['Рейтинг'] == 'Отклонение в пределах погрешности'].shape[0]
total_rows = df['Рейтинг'].notna().sum()
accuracy = (precision_rows / total_rows) * 100
print(precision_rows)
print(total_rows)
print(f"Точность прогноза по сравнению с фактом: {accuracy:.2f}%")

df['Рейтинг_план'] = df['Прирост_план'].apply(get_rating)

precision_rows = df[df['Рейтинг_план'] == 'Отклонение в пределах погрешности'].shape[0]
total_rows = df['Рейтинг_план'].notna().sum()
accuracy = (precision_rows / total_rows) * 100

print(f"Точность прогноза по сравнению с планом: {accuracy:.2f}%")

df['Рейтинг_прогноз'] = df['Прирост_прогноз'].apply(get_rating)

precision_rows = df[df['Рейтинг_прогноз'] == 'Отклонение в пределах погрешности'].shape[0]
total_rows = df['Рейтинг_прогноз'].notna().sum()
accuracy = (precision_rows / total_rows) * 100

print(f"Точность прогноза сравнению с прогнозом: {accuracy:.2f}%")
```
